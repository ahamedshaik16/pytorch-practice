# CODING TASK #6. PLOT HISTOGRAMS USING MATPLOTLIB

```python
mu = daily_returns_df['BTC'].mean()
sigma = daily_returns_df['BTC'].std()

num_bins = 40

plt.figure(figsize=(20,15))

daily_returns_df['BTC'].plot.hist(bins = 50, alpha = 0.5, figsize = (12, 6));

plt.grid()
plt.ylabel('Probability')
plt.title('Histogram: mu=' + str(mu) + ', sigma=' + str(sigma))
```
